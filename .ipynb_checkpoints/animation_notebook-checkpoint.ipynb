# Billboard Animation Tutorial: Hot 100 in 2025

This notebook is a beginner-friendly tutorial for making animated data visualizations with **Matplotlib** using **Billboard Hot 100** data.

For this version of the project, we focus on **2025**, which is the most recent complete year available in the dataset as of **April 24, 2026**. The dataset continues into 2026, but 2026 is only partial so far.

We build two examples:

1. A **bar chart race** showing songs building chart longevity over the course of 2025.
2. A **single-song trajectory** showing one song moving up and down the Hot 100 week by week.

The goal is not just to make something that moves, but to make something that is still readable, honest, and useful.


## Why this tutorial?

Many plotting tutorials stop at static charts. This one walks through an end-to-end animation workflow with:

- a public dataset,
- simple data cleaning,
- clear plotting code,
- and optional export steps for MP4 or GIF output.

This makes it a strong fit for the honors project because it teaches both a **visualization technique** and a **reproducible workflow**.


## Course connections

This project connects to several ideas from class:

- **Alberto Cairo** emphasizes that a visualization should be truthful, functional, and enlightening. The point of animation here is to clarify time-based change, not just to add motion.
- **Edward Tufte** focuses on graphical integrity and minimizing chartjunk. That is why these plots use simple styling, readable labels, and minimal decoration.
- **Segel & Heer** discuss narrative visualization and animated transitions. These animations guide the viewer through change over time and help preserve continuity from one frame to the next.
- **Cleveland & McGill** show that position and length on a common scale are accurate encodings. That makes bar charts and line plots good choices here.
- **Data Feminism** reminds us that every visualization reflects choices. This dataset highlights chart popularity, which is not the same as artistic quality or cultural importance.

One useful takeaway from class is that animation is not always the best answer. If exact comparison matters most, a static chart or small multiples may be clearer. Here, animation works well because the main story is about **movement over time**.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', 20)
```

## Load the Hot 100 dataset

This dataset is already in a tidy format, which makes it much easier to work with than the older Billboard dataset.

We still read it with **`latin-1` encoding** to avoid text issues.


```python
data_url = 'https://raw.githubusercontent.com/utdata/rwd-billboard-data/main/data-out/hot-100-current.csv'
local_fallback = 'hot-100-current.csv'

try:
    hot100 = pd.read_csv(data_url, encoding='latin-1')
    print('Loaded data from URL.')
except Exception as e:
    print(f'Could not load from URL: {e}')
    print('Trying local fallback file instead...')
    hot100 = pd.read_csv(local_fallback, encoding='latin-1')
    print('Loaded data from local file.')

hot100.head()
```

## Focus on 2025

The full dataset spans many decades, but for this tutorial we only want one year.

We convert `chart_week` to a real date, then filter to **2025**.


```python
hot100['chart_week'] = pd.to_datetime(hot100['chart_week'])
hot100_2025 = hot100[hot100['chart_week'].dt.year == 2025].copy()

hot100_2025['song'] = hot100_2025['title'] + ' - ' + hot100_2025['performer']
hot100_2025 = hot100_2025.sort_values(['chart_week', 'current_week']).reset_index(drop=True)

print('Rows in 2025 subset:', len(hot100_2025))
print('First week:', hot100_2025['chart_week'].min().date())
print('Last week:', hot100_2025['chart_week'].max().date())

hot100_2025.head()
```

## Get familiar with the columns

A few important columns:

- `chart_week`: the Billboard chart date
- `current_week`: the song's rank that week
- `last_week`: where it ranked the previous week
- `peak_pos`: the best rank the song has reached so far
- `wks_on_chart`: total weeks the song has spent on the chart so far

This is a nice dataset for animation because each row already represents **one song in one week**.


```python
print("Shape:", hot100_2025.shape)

print("\nColumns:")
print(hot100_2025.columns.tolist())
```

## Quick checks

Before animating, it helps to do a little exploratory analysis.


```python
song_summary = (
    hot100_2025.groupby('song')
    .agg(
        title=('title', 'first'),
        performer=('performer', 'first'),
        best_rank=('current_week', 'min'),
        longest_run_2025=('wks_on_chart', 'max'),
        first_2025_appearance=('chart_week', 'min'),
        last_2025_appearance=('chart_week', 'max'),
        weeks_present_in_2025=('chart_week', 'nunique')
    )
    .sort_values(['longest_run_2025', 'best_rank'], ascending=[False, True])
    .reset_index()
)

song_summary.head(10)
```

```python
top_10_longevity = song_summary.head(10).sort_values('longest_run_2025')

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top_10_longevity['song'], top_10_longevity['longest_run_2025'], color='steelblue')
ax.set_title('Top 10 Songs by Longest Hot 100 Run in 2025')
ax.set_xlabel('Weeks on chart (as recorded during 2025)')
ax.set_ylabel('Song')
plt.tight_layout()
plt.show()
```

## Prepare data for the bar chart race

For the bar chart race, we want the bars to grow over the year. We use the number of weeks each song has appeared in the Hot 100 **within 2025 so far**.

This creates a simple and easy-to-explain animation.


```python
race_df = hot100_2025[['chart_week', 'song']].copy()
race_df['weeks_in_2025_so_far'] = race_df.groupby('song').cumcount() + 1

# Keep a smaller set of songs so the chart stays readable.
top_songs = song_summary.head(15)['song']
race_top = race_df[race_df['song'].isin(top_songs)].copy()
all_dates = sorted(race_top['chart_week'].unique())

race_top.head()
```

## Animation 1: Bar chart race

This animation shows which songs build the longest presence on the Hot 100 throughout 2025.

To keep the chart readable, we show only the top 10 songs in each frame.


```python
fig, ax = plt.subplots(figsize=(12, 7))
colors = plt.cm.Blues(np.linspace(0.45, 0.9, 10))


def draw_barchart(current_date):
    ax.clear()

    current = race_top[race_top['chart_week'] == current_date].copy()
    current = current.sort_values('weeks_in_2025_so_far', ascending=False).head(10)
    current = current.sort_values('weeks_in_2025_so_far')

    ax.barh(current['song'], current['weeks_in_2025_so_far'], color=colors[:len(current)])
    ax.set_xlim(0, race_top['weeks_in_2025_so_far'].max() + 2)
    ax.set_xlabel('Weeks appearing in the Hot 100 during 2025 so far')
    ax.set_ylabel('Song')
    ax.set_title(
    "Billboard Hot 100 Bar Chart Race\n"
    + "2025 song longevity over time | "
    + str(pd.to_datetime(current_date).date())
)

    for i, value in enumerate(current['weeks_in_2025_so_far']):
        ax.text(value + 0.3, i, str(value), va='center', fontsize=10)

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.tight_layout()


bar_chart_race = FuncAnimation(fig, draw_barchart, frames=all_dates, interval=250, repeat=False)
HTML(bar_chart_race.to_jshtml())
```

## Pick one song for a trajectory animation

For the second example, we focus on a single song. This creates a more personal and easier-to-follow story.

The chart below lets us choose one song from the strongest performers in 2025.


```python
song_summary[['song', 'best_rank', 'weeks_present_in_2025']].head(15)
```

For now, I am choosing **"Luther - Kendrick Lamar & SZA"** as an example. If that song is not present for any reason, the notebook will automatically fall back to the first song in `song_summary`.


```python
preferred_song = 'Luther - Kendrick Lamar & SZA'

if preferred_song in set(hot100_2025['song']):
    song_title = preferred_song
else:
    song_title = song_summary.loc[0, 'song']

song_path = hot100_2025[hot100_2025['song'] == song_title].sort_values('chart_week').copy()
print('Chosen song:', song_title)
print('Rows for chosen song:', len(song_path))

song_path[['chart_week', 'current_week', 'peak_pos', 'wks_on_chart']].head()
```

## Animation 2: One song's rank trajectory

A lower chart rank is better, so we reverse the y-axis so rank **1** appears at the top.


```python
if song_path.empty:
    print('No rows found for the chosen song.')
else:
    fig, ax = plt.subplots(figsize=(10, 5))

    def draw_song_path(frame):
        ax.clear()
        current = song_path.iloc[:frame + 1]

        ax.plot(current['chart_week'], current['current_week'], color='darkred', linewidth=2.5)
        ax.scatter(current['chart_week'].iloc[-1], current['current_week'].iloc[-1], color='black', s=50)

        ax.set_ylim(100, 1)
        ax.set_title(f'Hot 100 Rank Trajectory for {song_title}')
        ax.set_xlabel('Chart week')
        ax.set_ylabel('Hot 100 rank')
        ax.grid(True, alpha=0.3)

        latest_rank = current['current_week'].iloc[-1]
        latest_date = current['chart_week'].iloc[-1].date()
        ax.text(
            0.02,
            0.05,
            f'Latest point: rank {latest_rank} on {latest_date}',
            transform=ax.transAxes,
            fontsize=10,
            bbox=dict(facecolor='white', alpha=0.8, edgecolor='none')
        )

        plt.tight_layout()

    song_animation = FuncAnimation(fig, draw_song_path, frames=len(song_path), interval=250, repeat=False)
    display(HTML(song_animation.to_jshtml()))
```
